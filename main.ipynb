```python
from sklearn.preprocessing import OneHotEncoder

oenc = OneHotEncoder(sparse_output=False)
encoded_type = oenc.fit_transform(X[["type"]])

X[oenc.get_feature_names_out(["type"])] = encoded_type

X = X.drop("type", axis=1)
```

```python
X.info()
```

```python
X
```

```python
X = X.drop(columns=["nameOrig","nameDest"], axis=1)
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size = 0.2,
    random_state = 42,
    stratify = Y
)
```

```python
x_train
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

x_train_scaled =  scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier
```

```python
'''grid = {
    "max_depth": [5, 10, 15, 20, 30],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 5, 10],
    "min_impurity_decrease": [0.0, 0.01, 0.05]
}


model = DecisionTreeClassifier()

grid_model = GridSearchCV(
    estimator=model,
    param_grid=grid,
    cv=5,
    scoring="accuracy",
    n_jobs=1
)

grid_model.fit(x_train, y_train)
print(grid_model.best_params_)
print(grid_model.best_score_)'''
```

```python
model1 = DecisionTreeClassifier()
model1.fit(x_train_scaled, y_train)
```

```python
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix
```

```python
y_pred = model1.predict(x_test)

print(accuracy_score(y_test, y_pred))
```

```python
model = DecisionTreeClassifier(
    max_depth = 15,
    min_samples_split=5,
    min_samples_leaf=2,
    min_impurity_decrease=0.02
)

model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)

print(accuracy_score(y_test, y_pred))
print(f1_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
# Overfitting check
y_pred_train = model.predict(x_train)

print(accuracy_score(y_train, y_pred_train))
print(f1_score(y_train, y_pred_train))
```

```python
print(pd.Series(y_pred_train).value_counts())
```

```python
model = DecisionTreeClassifier(
    class_weight="balanced",
    random_state=42
)

model.fit(x_train, y_train)

y_pred_train = model.predict(x_train)

print("Accuracy:", accuracy_score(y_train, y_pred_train))
print("F1:", f1_score(y_train, y_pred_train))
print(confusion_matrix(y_train, y_pred_train))
```

# Final Decision Tree with Pruning Accuracy was - 99.87 % (Overfit)

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

lr.fit(x_train_scaled, y_train)

y_pred_lr = lr.predict(x_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print("F1:", f1_score(y_test, y_pred_lr))
```
